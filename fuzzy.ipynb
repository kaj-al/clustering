```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
```

```python
data = np.array([
    [1, 2],
    [1.5, 1.8],
    [2, 2.5],
    [8, 8],
    [8.5, 9],
    [9, 8.5],
    [5, 5],
    [5.5, 5.2]
])
```

```python
data = data.T

n_clusters = 3

centers, membership, _, _, _, _, _ = fuzz.cluster.cmeans(
    data,
    c=n_clusters,
    m=2,
    error=0.005,
    maxiter=1000,
    init=None
)

print("Cluster Centers:")
print(centers)

print("\nMembership Values:")
print(membership)

# Assign each point to cluster with highest membership value
labels = np.argmax(membership, axis=0)

print("\nCluster Labels:")
print(labels)
```

```python
plt.figure(figsize=(8, 6))

for i in range(n_clusters):
    points = data[:, labels == i]

    plt.scatter(
        points[0],
        points[1],
        label=f"Cluster {i + 1}"
    )

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker="X",
    s=200,
    label="Centers"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Fuzzy C-Means Clustering")
plt.legend()
plt.show()
```
